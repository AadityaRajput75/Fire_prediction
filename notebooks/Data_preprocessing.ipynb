```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
```

```python
df.head()
```

```python
df.info()
```

Check Missing Values


```python
df[df.isnull().any(axis=1)]
```

```python
df.loc[:122,"Region"]=0
df.loc[122:,"Region"]=1
```

```python
df['Region']=df['Region'].astype(int)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df=df.dropna().reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
#fix column spaces=
df.columns=df.columns.str.strip()
df.columns
```

```python
##change the required columns as integer datatype
df[['month','day','year','Temperature','RH','Ws']]=df[['month','day','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

```python
#changing other to float datatype
objects=[features for features in df.columns if df[features].dtypes=='string']
objects
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
```

```python
df.info()
```

```python
#Save the clean dataset
df.to_csv('Algerian_forest_fires_cleaned.csv',index=False)
```

Exploratory Data Analysis


```python
df_copy=df.drop(['day','month','year'],axis=1)
```

Encoding categories in classes

```python
df_copy['Classes']=np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
#categories in classes
df_copy['Classes'].value_counts()
```

```python
#Density plot for all features
plt.style.use('seaborn-v0_8')
df_copy.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
#percentage for pie chart
percentage=df_copy['Classes'].value_counts(normalize=True)*100
```

```python
#plotting pie chart
classlabels=["Fire","Not fire"]
plt.figure(figsize=(12,7))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%')
```

```python
#correlation
df_copy.corr()
```
